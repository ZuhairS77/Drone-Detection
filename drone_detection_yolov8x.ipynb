## Train

The dataset must follow the layout documented in `README.md`. Ultralytics downloads the starting weights when needed.

```python
import subprocess
import sys

subprocess.run(
    [sys.executable, "drone_detection_yolov8x.py", "train", "--data", "coco.yaml"],
    check=True,
)
```

## Track a video

After training, set `source` to a local video path or stream URL. Add `--show` to display the tracking window or `--save` to save annotated results.

```python
source = "path/to/video.mp4"
weights = "runs/detect/train/weights/best.pt"

subprocess.run(
    [
        sys.executable,
        "drone_detection_yolov8x.py",
        "track",
        "--weights",
        weights,
        "--source",
        source,
        "--show",
    ],
    check=True,
)
```
